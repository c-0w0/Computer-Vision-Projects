```python
# General import
import os

# 3rd-party import
import cv2
import matplotlib.pyplot as plt

# Local import
```

```python
img_path = '../data/cat.jpg'
img = cv2.imread(img_path)
```

```python
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

```python
plt.imshow(img_rgb)
plt.title("Image")
plt.axis('off')
plt.show()
```

```python
k_size = 7 # kernel size
img_mean_blur = cv2.blur(img_rgb, (k_size, k_size))
img_gaussian_blur = cv2.GaussianBlur(img_rgb, (k_size, k_size), 5)
img_median_blur = cv2.medianBlur(img_rgb, k_size)
```

```python
fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(15, 5))

ax1.imshow(img_mean_blur)
ax1.set_title("Blur with rectangle")
ax1.axis('off')

ax2.imshow(img_gaussian_blur)
ax2.set_title("Gaussian blur")
ax2.axis('off')

ax3.imshow(img_median_blur)
ax3.set_title("Median Blur")
ax3.axis('off')

plt.tight_layout()
plt.show()
```
