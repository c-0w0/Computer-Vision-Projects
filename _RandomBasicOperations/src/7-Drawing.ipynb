```python
# General import
import os

# 3rd-party import
import cv2
import matplotlib.pyplot as plt

# Local import
```

```python
img_path = '../data/cat.jpg'
img = cv2.imread(img_path)
```

```python
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# plt.figure(figsize=(8, 6))
plt.imshow(img_rgb)
plt.title("RGB Image")
plt.axis('off')
plt.show()
```

```python
# Drawing a line
cv2.line(
    img_rgb,       # img
    (100, 150),    # pos1
    (300, 450),    # pos2
    (0, 255, 0),   # color
    3              # thickness
)

plt.imshow(img_rgb)
plt.axis('off')
plt.show()
```

```python
# Drawing a circle
cv2.circle(
    img_rgb,        # img
    (200, 150),     # center pos
    50,             # radius
    (255, 0, 0),    # color
    4               # thickness
)

plt.imshow(img_rgb)
plt.axis('off')
plt.show()
```

```python
# Drawing text
cv2.putText(
    img_rgb,                    # img
    'Meow!',                 # text
    (1000, 150),                 # pos
    cv2.FONT_HERSHEY_SIMPLEX,   # font
    2,                          # text size
    (255, 0, 0),                # color
    10                          # thickness
)

plt.imshow(img_rgb)
plt.axis('off')
plt.show()
```
