```python
# General import
import os

# 3rd-party import
import cv2
import matplotlib.pyplot as plt

# Local import
```

```python
img_path = '../data/cat.jpg'
img = cv2.imread(img_path)
```

```python
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
img_gray = cv2.cvtColor(img_rgb, cv2.COLOR_BGR2GRAY)
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 5))

ax1.imshow(img_rgb)
ax1.set_title("RGB Image")
ax1.axis('off')

ax2.imshow(img_gray, cmap='gray')
ax2.set_title("Grayscale image")
ax2.axis('off')

plt.tight_layout()
plt.show()
```

## Thresholding

```python
ret, thresh1 = cv2.threshold(img_gray, 50, 255, cv2.THRESH_BINARY)
ret, thresh2 = cv2.threshold(img_gray, 100, 255, cv2.THRESH_BINARY)
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 5))

ax1.imshow(thresh1, cmap='gray')
ax1.set_title("Threshold=50")
ax1.axis('off')

ax2.imshow(thresh2, cmap='gray')
ax2.set_title("Threshold=100")
ax2.axis('off')

plt.tight_layout()
plt.show()
```

## Adaptive threshold

```python
img_path = '../data/writing.png'
img = cv2.imread(img_path)
```

```python
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 10))

ax1.imshow(img_rgb)
ax1.set_title("RGB Image")
ax1.axis('off')

ax2.imshow(img_gray, cmap='gray')
ax2.set_title("Grayscale image")
ax2.axis('off')

plt.tight_layout()
plt.show()
```

```python
adaptive_thresh = cv2.adaptiveThreshold(img_gray, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY, 21, 30)
```

```python
plt.figure(figsize=(10, 10))
plt.imshow(adaptive_thresh, cmap='gray')
plt.title("Adaptive thresh")
plt.axis('off')
plt.show()
```
