```python
# General import
import os

# 3rd-party import
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Local import
```

```python
img_path = '../data/cat.jpg'
img = cv2.imread(img_path)
```

```python
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# plt.figure(figsize=(8, 6))
plt.imshow(img_rgb)
plt.title("Image")
plt.axis('off')
plt.show()
```

```python
img_edge = cv2.Canny(img, 100, 200)
img_edge_d = cv2.dilate(img_edge, np.ones((3, 3), dtype=np.int8))
img_edge_e = cv2.erode(img_edge_d, np.ones((3, 3), dtype=np.int8))
```

```python
fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(15, 5))

ax1.imshow(img_edge, cmap='gray')
ax1.set_title("Canny edge detection")
ax1.axis('off')

ax2.imshow(img_edge_d, cmap='gray')
ax2.set_title("Dilation")
ax2.axis('off')

ax3.imshow(img_edge_e, cmap='gray')
ax3.set_title("Erosion")
ax3.axis('off')

plt.tight_layout()
plt.show()
```
